````python
import pandas as pd

df = pd.read_csv('variant_summary.txt.gz', sep='\t', low_memory=False)

# Step 3a â€” baseline filters first (eliminate junk quickly)
df = df[
    (df['Assembly'] == 'GRCh38') &
    (df['Type'] == 'single nucleotide variant') &
    (df['Origin'].str.contains('germline', case=False, na=False))
]

# Step 3b â€” keep only Expert Panel and Practice Guideline (3â˜…/4â˜…)
expert_statuses = ['reviewed by expert panel', 'practice guideline']
df_expert = df[df['ReviewStatus'].isin(expert_statuses)]

# Step 3c â€” Pathogenic class: P and LP, but EXCLUDE anything
# that also mentions Benign or Uncertain in the same record
pathogenic = df_expert[
    df_expert['ClinicalSignificance'].str.contains(
        r'Pathogenic|Likely pathogenic', case=False, na=False
    ) &
    ~df_expert['ClinicalSignificance'].str.contains(
        r'Benign|Uncertain|Conflicting', case=False, na=False
    )
].copy()
pathogenic['label'] = 1

print(f"Raw pathogenic variants: {len(pathogenic)}")
print(pathogenic['ClinicalSignificance'].value_counts())
````

````python
# Extract all genes present in pathogenic set
all_genes = (pathogenic['GeneSymbol']
             .str.split(';').explode()
             .str.strip()
             .dropna()
             .unique()
             .tolist())

# Hardcode the disease-specific panels but VERIFY they exist in all_genes
cancer_panel = ['BRCA1','BRCA2','TP53','MLH1','MSH2','MSH6',
                'PMS2','APC','CDKN2A','CDH1','PTEN','STK11',
                'PALB2','CHEK2','ATM','BRIP1','RAD51C','RAD51D']

pah_panel  = ['PAH']
cftr_panel = ['CFTR']

# Sanity check — are these genes actually in our data?
for gene in cancer_panel + pah_panel + cftr_panel:
    if gene not in all_genes:
        print(f"WARNING: {gene} not found in ClinVar pathogenic set")
````

````python
benign_clinvar = df_expert[
    df_expert['ClinicalSignificance'].str.contains(
        r'Benign|Likely benign', case=False, na=False
    ) &
    ~df_expert['ClinicalSignificance'].str.contains(
        r'Pathogenic|Uncertain|Conflicting', case=False, na=False
    ) &
    df_expert['GeneSymbol'].isin(all_genes)  # same gene scope as pathogenic
].copy()
benign_clinvar['label'] = 0
benign_clinvar['source'] = 'ClinVar'

print(f"ClinVar benign variants: {len(benign_clinvar)}")
gap = len(pathogenic) - len(benign_clinvar)
print(f"Gap to fill from gnomAD: {gap} variants")
````

````python
# Check if these genes exist in ClinVar at ALL (any review status)
missing_genes = ['CDKN2A','STK11','CHEK2','BRIP1','RAD51C','RAD51D']

# Load unfiltered df (before review status filter)
df_all_snv = df[df['Type'] == 'single nucleotide variant']

for gene in missing_genes:
    subset = df_all_snv[
        df_all_snv['GeneSymbol'].str.contains(gene, na=False)
    ]
    p_subset = subset[subset['ClinicalSignificance'].str.contains(
        'Pathogenic', case=False, na=False
    )]
    star_counts = p_subset['ReviewStatus'].value_counts()
    print(f"\n{gene}: {len(p_subset)} pathogenic entries")
    print(star_counts.to_string())
````

````python
import re

# CELL 10 — Keep only what matters, drop the rest
cols_to_keep = [
    'AlleleID', 'Type', 'Name', 'GeneID', 'GeneSymbol',
    'ClinicalSignificance', 'ReviewStatus', 'RS# (dbSNP)',
    'Assembly', 'Chromosome', 'Start', 'Stop', 'Origin',
    'ReferenceAlleleVCF', 'AlternateAlleleVCF'
]

# Keep only existing columns (some may differ by ClinVar version)
cols_to_keep = [c for c in cols_to_keep if c in df.columns]
df_slim = df[cols_to_keep].copy()

print(f"Columns kept: {len(cols_to_keep)}")
print(f"Columns dropped: {len(df.columns) - len(cols_to_keep)}")
print(f"Memory before: {df.memory_usage(deep=True).sum() / 1e6:.1f} MB")
print(f"Memory after:  {df_slim.memory_usage(deep=True).sum() / 1e6:.1f} MB")


# CELL 11 — Parse amino acid change from the Name column
# e.g. "NM_000410.4(HFE):c.845G>A (p.Cys282Tyr)" → ref_aa=Cys, alt_aa=Tyr, pos=282

def parse_hgvs_protein(name):
    """Extract ref AA, alt AA, protein position from HGVS Name field."""
    match = re.search(r'p\.([A-Za-z]{3})(\d+)([A-Za-z]{3})', str(name))
    if match:
        return match.group(1), match.group(2), match.group(3)
    return None, None, None

def parse_hgvs_cdna(name):
    """Extract ref nucleotide, alt nucleotide from cDNA change."""
    match = re.search(r'c\.\d+([A-Z])>([A-Z])', str(name))
    if match:
        return match.group(1), match.group(2)
    return None, None

df_slim[['ref_aa', 'protein_pos', 'alt_aa']] = df_slim['Name'].apply(
    lambda x: pd.Series(parse_hgvs_protein(x))
)
df_slim[['ref_nuc_cdna', 'alt_nuc_cdna']] = df_slim['Name'].apply(
    lambda x: pd.Series(parse_hgvs_cdna(x))
)

# Quick check
print("\nSample parsed fields:")
print(df_slim[['Name','ref_aa','protein_pos',
               'alt_aa','ref_nuc_cdna','alt_nuc_cdna']].head(10))

# Check parse success rate
success = df_slim['ref_aa'].notna().mean()
print(f"\nHGVS parse success rate: {success:.1%}")


# CELL 12 — Build join key for dbNSFP (needed in annotation step)
# Format: chr:pos:ref:alt  — matches dbNSFP coordinate format
df_slim['join_key'] = (
    df_slim['Chromosome'].astype(str) + ':' +
    df_slim['Start'].astype(str) + ':' +
    df_slim['ReferenceAlleleVCF'].astype(str) + ':' +
    df_slim['AlternateAlleleVCF'].astype(str)
)

# Verify no duplicate join keys within the same label class
dupes = df_slim.groupby(['join_key', 'label']).size()
dupes_problem = dupes[dupes > 1]
if len(dupes_problem) > 0:
    print(f"⚠️  {len(dupes_problem)} duplicate join keys — will deduplicate")
else:
    print("✅ No duplicate join keys")

print(f"\nFinal slim dataset: {df_slim.shape}")
print(df_slim[['join_key','ref_aa','alt_aa','label']].head())


## What the HGVS parse gives you

# From the `Name` column alone you can extract:
# ```
# NM_000410.4(HFE):c.845G>A (p.Cys282Tyr)
#                     │  │ │    │         │
#          ref_nuc ───┘  │ └── alt_nuc   │
#          cDNA pos ─────┘              │
#                              ref_aa ──┘─► Cys (Cysteine)
#                              protein_pos ─► 282
#                              alt_aa ──────► Tyr (Tyrosine)
````
