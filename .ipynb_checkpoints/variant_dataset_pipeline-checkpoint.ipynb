# Genomic Variant Dataset Collection & Feature Engineering Pipeline
**Competition: Pathogenic vs Benign Missense Variant Classification**

This notebook collects, processes, and cleans variant data from ClinVar and gnomAD according to ACMG guidelines.  
Features are engineered to cover: sequence context, biochemical effects, evolutionary conservation, population data, and in silico scores.

---
**Datasets produced:**
- General Dataset (1500P / 1500B train | 1000P / 1000B test)
- Hereditary Cancer Panel (200/200 | 100/100)
- PAH Gene Panel (200/200 | 100/100)
- CFTR Gene Panel (70/70 | 30/30)

## 1. Reference Lookups & Constants

```python
# ── Amino-acid physicochemical property tables ──────────────────────────────
AA_HYDROPHOBICITY = {  # Kyte-Doolittle scale
    'A': 1.8,  'R': -4.5, 'N': -3.5, 'D': -3.5, 'C': 2.5,
    'Q': -3.5, 'E': -3.5, 'G': -0.4, 'H': -3.2, 'I': 4.5,
    'L': 3.8,  'K': -3.9, 'M': 1.9,  'F': 2.8,  'P': -1.6,
    'S': -0.8, 'T': -0.7, 'W': -0.9, 'Y': -1.3, 'V': 4.2,
    'X': 0.0
}

AA_POLARITY = {  # Zimmermann scale (1 = polar, 0 = non-polar)
    'A': 0, 'R': 1, 'N': 1, 'D': 1, 'C': 0,
    'Q': 1, 'E': 1, 'G': 0, 'H': 1, 'I': 0,
    'L': 0, 'K': 1, 'M': 0, 'F': 0, 'P': 0,
    'S': 1, 'T': 1, 'W': 0, 'Y': 1, 'V': 0,
    'X': 0
}

AA_MW = {  # molecular weight (Da)
    'A': 89.1,  'R': 174.2, 'N': 132.1, 'D': 133.1, 'C': 121.2,
    'Q': 146.2, 'E': 147.1, 'G': 75.0,  'H': 155.2, 'I': 131.2,
    'L': 131.2, 'K': 146.2, 'M': 149.2, 'F': 165.2, 'P': 115.1,
    'S': 105.1, 'T': 119.1, 'W': 204.2, 'Y': 181.2, 'V': 117.1,
    'X': 110.0
}

AA_CHARGE = {  # charge at pH 7.4
    'D': -1, 'E': -1, 'K': 1, 'R': 1, 'H': 0.1,
    'A': 0,  'C': 0,  'F': 0, 'G': 0, 'I': 0,
    'L': 0,  'M': 0,  'N': 0, 'P': 0, 'Q': 0,
    'S': 0,  'T': 0,  'V': 0, 'W': 0, 'Y': 0, 'X': 0
}

# IUPAC one-to-three letter code
AA1TO3 = {
    'A':'Ala','R':'Arg','N':'Asn','D':'Asp','C':'Cys',
    'Q':'Gln','E':'Glu','G':'Gly','H':'His','I':'Ile',
    'L':'Leu','K':'Lys','M':'Met','F':'Phe','P':'Pro',
    'S':'Ser','T':'Thr','W':'Trp','Y':'Tyr','V':'Val',
    'X':'Xaa'
}

# Gene panels used in the competition
HEREDITARY_CANCER_GENES = [
    'BRCA1','BRCA2','TP53','MLH1','MSH2','MSH6','PMS2','APC',
    'PALB2','CHEK2','ATM','RAD51C','RAD51D','BARD1','NBN',
    'PTEN','STK11','CDH1','VHL','RET','MEN1','SDHB','SDHD'
]

PAH_GENES  = ['PAH']
CFTR_GENES = ['CFTR']

print("Constants loaded.")
```

## 2. Data Collection — ClinVar (Pathogenic Variants)

Download and parse ClinVar VCF / variant_summary for missense variants with  
**Expert Panel / Practice Guideline** review status (3★ / 4★).

```python
# ── ClinVar variant_summary download ────────────────────────────────────────
CLINVAR_SUMMARY_URL = (
    "https://ftp.ncbi.nlm.nih.gov/pub/clinvar/tab_delimited/"
    "variant_summary.txt.gz"
)

CLINVAR_LOCAL = RAW_DIR / "variant_summary.txt.gz"

def download_file(url: str, dest: Path, chunk_size: int = 1 << 20):
    """Stream-download a large file with a progress bar."""
    if dest.exists():
        print(f"[SKIP] {dest.name} already downloaded.")
        return
    print(f"Downloading {url} …")
    with requests.get(url, stream=True, timeout=120) as r:
        r.raise_for_status()
        total = int(r.headers.get('content-length', 0))
        with open(dest, 'wb') as f, tqdm(total=total, unit='B',
                                          unit_scale=True) as bar:
            for chunk in r.iter_content(chunk_size):
                f.write(chunk)
                bar.update(len(chunk))
    print(f"Saved → {dest}")

download_file(CLINVAR_SUMMARY_URL, CLINVAR_LOCAL)
```

```python
# ── Parse ClinVar variant_summary ───────────────────────────────────────────
REVIEW_STATUS_KEEP = {
    'reviewed by expert panel',
    'practice guideline'
}

CLINSIG_PATHOGENIC = {'Pathogenic', 'Likely pathogenic',
                       'Pathogenic/Likely pathogenic'}
CLINSIG_BENIGN     = {'Benign', 'Likely benign',
                       'Benign/Likely benign'}

def load_clinvar(path: Path) -> pd.DataFrame:
    print("Reading ClinVar variant_summary …")
    cols = [
        '#AlleleID','Type','Name','GeneSymbol',
        'ClinicalSignificance','ReviewStatus',
        'ReferenceAllele','AlternateAllele',
        'Chromosome','Start','PositionVCF',
        'ReferenceAlleleVCF','AlternateAlleleVCF',
        'NumberSubmitters','LastEvaluated',
        'RS# (dbSNP)','nsv/esv (dbVar)','RCVaccession',
        'PhenotypeIDS','PhenotypeList','Origin',
        'OriginSimple','Assembly','ChromosomeAccession',
        'Stop','Cytogenetics','ReviewStatus',
        'NumberSubmitters','Guidelines','TestedInGTR',
        'OtherIDs','SubmitterCategories','VariationID'
    ]
    # Read only relevant columns to save memory
    use_cols = [
        '#AlleleID','Type','Name','GeneSymbol',
        'ClinicalSignificance','ReviewStatus',
        'ReferenceAllele','AlternateAllele',
        'Assembly'
    ]
    df = pd.read_csv(
        path, sep='\t', compression='gzip',
        usecols=lambda c: c in use_cols,
        low_memory=False
    )
    df.columns = df.columns.str.strip().str.lstrip('#')
    df.rename(columns={'AlleleID': 'allele_id'}, inplace=True)

    # ── Filters ─────────────────────────────────────────────────────────────
    # 1. Missense variants only
    df = df[df['Type'].str.lower().str.contains('single nucleotide', na=False)]
    # 2. GRCh38 assembly
    df = df[df['Assembly'] == 'GRCh38']
    # 3. Expert Panel or Practice Guideline
    df = df[df['ReviewStatus'].str.lower().isin(REVIEW_STATUS_KEEP)]
    # 4. Pathogenic or Benign significance only
    path_mask  = df['ClinicalSignificance'].isin(CLINSIG_PATHOGENIC)
    benign_mask = df['ClinicalSignificance'].isin(CLINSIG_BENIGN)
    df = df[path_mask | benign_mask].copy()

    # ── Label encoding ──────────────────────────────────────────────────────
    df['label'] = np.where(path_mask[df.index], 1, 0)

    # ── Extract HGVS-p change from 'Name' field ──────────────────────────────
    # ClinVar Name format: NM_xxxx.x(GENE):c.xxx p.(AAAposAAA)
    df['hgvs_p'] = df['Name'].str.extract(r'p\.\(?(\w+\d+\w+)\)?')

    # Extract codon change: c.xxx
    df['hgvs_c'] = df['Name'].str.extract(r'(c\.[^\s\(]+)')

    print(f"ClinVar records loaded: {len(df):,}  "
          f"(Pathogenic={path_mask.sum():,}, Benign={benign_mask.sum():,})")
    return df

cv_df = load_clinvar(CLINVAR_LOCAL)
```

## 3. Data Collection — gnomAD (Benign Population Variants)

Fetch high-frequency missense variants (AF > 0.01) from gnomAD v4 for genes  
present in the ClinVar data to balance the Benign class.

```python
# ── gnomAD GraphQL API helper ────────────────────────────────────────────────
GNOMAD_API = "https://gnomad.broadinstitute.org/api"

GNOMAD_QUERY = """
query GeneVariants($geneId: String!, $dataset: DatasetId!) {
  gene(gene_symbol: $geneId, reference_genome: GRCh38) {
    variants(dataset: $dataset) {
      variant_id
      consequence
      hgvsc
      hgvsp
      exome { ac af an }
      genome { ac af an }
      transcript_consequence {
        amino_acids
        codons
        sift_prediction  sift_score
        polyphen_prediction polyphen_score
      }
      in_silico_predictors { id  value  flag }
    }
  }
}
"""

def fetch_gnomad_gene(gene_symbol: str,
                      dataset: str = "gnomad_r4",
                      min_af: float = 0.01,
                      max_retries: int = 3) -> pd.DataFrame:
    """Fetch high-frequency missense variants for one gene from gnomAD."""
    records = []
    for attempt in range(max_retries):
        try:
            resp = requests.post(
                GNOMAD_API,
                json={'query': GNOMAD_QUERY,
                      'variables': {'geneId': gene_symbol, 'dataset': dataset}},
                timeout=60
            )
            resp.raise_for_status()
            data = resp.json()
            variants = (data.get('data', {}).get('gene') or {}).get('variants', [])
            break
        except Exception as e:
            print(f"  Retry {attempt+1}/{max_retries} for {gene_symbol}: {e}")
            time.sleep(2 ** attempt)
    else:
        print(f"  [FAIL] Could not fetch {gene_symbol}")
        return pd.DataFrame()

    for v in variants:
        if v.get('consequence') != 'missense_variant':
            continue
        # Use genome AF; fall back to exome
        af = (v.get('genome') or {}).get('af') or \
             (v.get('exome') or {}).get('af') or 0
        if af < min_af:
            continue
        tc = v.get('transcript_consequence') or {}
        row = {
            'gene':            gene_symbol,
            'variant_id':      v.get('variant_id'),
            'hgvs_c':          v.get('hgvsc'),
            'hgvs_p':          v.get('hgvsp'),
            'af_gnomad':       af,
            'ac_gnomad':       (v.get('genome') or {}).get('ac'),
            'an_gnomad':       (v.get('genome') or {}).get('an'),
            'amino_acids':     tc.get('amino_acids'),
            'codons':          tc.get('codons'),
            'sift_pred':       tc.get('sift_prediction'),
            'sift_score':      tc.get('sift_score'),
            'polyphen_pred':   tc.get('polyphen_prediction'),
            'polyphen_score':  tc.get('polyphen_score'),
            'label':           0   # Benign
        }
        # Flatten in-silico predictors
        for pred in (v.get('in_silico_predictors') or []):
            row[f"insilico_{pred['id']}"] = pred.get('value')
        records.append(row)

    return pd.DataFrame(records)

print("gnomAD helper defined.")
```

```python
# ── Collect gnomAD benign variants for all panel genes ───────────────────────
all_genes = list(set(HEREDITARY_CANCER_GENES + PAH_GENES + CFTR_GENES +
                     cv_df['GeneSymbol'].dropna().unique().tolist()))

gnomad_frames = []
for gene in tqdm(all_genes, desc="gnomAD genes"):
    gdf = fetch_gnomad_gene(gene)
    if not gdf.empty:
        gdf['GeneSymbol'] = gene
        gnomad_frames.append(gdf)
    time.sleep(0.5)   # be polite to the API

gnomad_df = pd.concat(gnomad_frames, ignore_index=True) if gnomad_frames else pd.DataFrame()
print(f"gnomAD benign variants collected: {len(gnomad_df):,}")
gnomad_df.to_parquet(RAW_DIR / "gnomad_benign_raw.parquet", index=False)
```

## 4. Merge & Deduplicate ClinVar + gnomAD

```python
# ── Harmonise column names ───────────────────────────────────────────────────
cv_df.rename(columns={
    'GeneSymbol':  'gene',
    'ReferenceAllele': 'ref',
    'AlternateAllele': 'alt'
}, inplace=True)

# Parse ref/alt from HGVS_p in gnomAD
def parse_aa_change(hgvs_p: str):
    """Return (ref_aa, pos, alt_aa) from HGVS protein notation."""
    if not isinstance(hgvs_p, str):
        return None, None, None
    m = re.search(r'p\.([A-Za-z]{3})(\d+)([A-Za-z]{3})', hgvs_p)
    if m:
        return m.group(1), int(m.group(2)), m.group(3)
    return None, None, None

# Create a stable dedup key = gene + hgvs_p
for df in [cv_df, gnomad_df]:
    df['dedup_key'] = df['gene'].fillna('') + ':' + df['hgvs_p'].fillna('')

combined = pd.concat([cv_df, gnomad_df], ignore_index=True, sort=False)

# Drop exact duplicates (same gene + HGVS_p), keep ClinVar label
combined.drop_duplicates(subset='dedup_key', keep='first', inplace=True)
combined.reset_index(drop=True, inplace=True)

print(f"Combined variants: {len(combined):,}")
print(combined['label'].value_counts().rename({0:'Benign', 1:'Pathogenic'}))
```

## 5. Feature Engineering

Build the full feature matrix according to the competition specification.

### 5.1 Sequence & Change Information

```python
def parse_codon_change(hgvs_c: str):
    """
    Extract ref nucleotide, alt nucleotide, and position from HGVS coding.
    e.g. c.1234A>G  → ref_nuc=A, alt_nuc=G, cds_pos=1234
    """
    if not isinstance(hgvs_c, str):
        return pd.Series({'ref_nuc': None, 'alt_nuc': None, 'cds_pos': None})
    m = re.search(r'c\.(\d+)([ACGT])>([ACGT])', hgvs_c)
    if m:
        return pd.Series({
            'ref_nuc': m.group(2),
            'alt_nuc': m.group(3),
            'cds_pos': int(m.group(1))
        })
    return pd.Series({'ref_nuc': None, 'alt_nuc': None, 'cds_pos': None})

nuc_feats = combined['hgvs_c'].apply(parse_codon_change)
combined = pd.concat([combined, nuc_feats], axis=1)

# ── Amino-acid change ────────────────────────────────────────────────────────
def parse_aa_features(hgvs_p: str):
    """
    Returns ref_aa (3-letter), alt_aa (3-letter), protein_pos.
    Normalises to 1-letter internally.
    """
    THREE_TO_ONE = {v: k for k, v in AA1TO3.items()}
    if not isinstance(hgvs_p, str):
        return pd.Series({'ref_aa3': None, 'alt_aa3': None,
                          'ref_aa1': None, 'alt_aa1': None,
                          'protein_pos': None})
    m = re.search(r'p\.([A-Za-z]{3})(\d+)([A-Za-z]{3})', hgvs_p)
    if m:
        r3, pos, a3 = m.group(1).capitalize(), int(m.group(2)), m.group(3).capitalize()
        return pd.Series({
            'ref_aa3':    r3,
            'alt_aa3':    a3,
            'ref_aa1':    THREE_TO_ONE.get(r3, 'X'),
            'alt_aa1':    THREE_TO_ONE.get(a3, 'X'),
            'protein_pos': pos
        })
    return pd.Series({'ref_aa3': None, 'alt_aa3': None,
                      'ref_aa1': None, 'alt_aa1': None,
                      'protein_pos': None})

aa_feats = combined['hgvs_p'].apply(parse_aa_features)
combined = pd.concat([combined, aa_feats], axis=1)

print("Sequence & change features added.")
combined[['ref_nuc','alt_nuc','cds_pos',
          'ref_aa3','alt_aa3','protein_pos']].head()
```

### 5.2 Genomic & Proteomic Neighbourhood (±5 context)

```python
# ── Ensembl REST API: fetch CDS sequence for a transcript ───────────────────
ENSEMBL_REST = "https://rest.ensembl.org"

def get_transcript_cds(transcript_id: str, timeout: int = 15) -> str:
    """Return CDS sequence for an Ensembl transcript."""
    url = f"{ENSEMBL_REST}/sequence/id/{transcript_id}"
    params = {'type': 'cds', 'content-type': 'text/plain'}
    try:
        r = requests.get(url, params=params, timeout=timeout)
        r.raise_for_status()
        return r.text.strip()
    except Exception:
        return ''

def get_transcript_protein(transcript_id: str, timeout: int = 15) -> str:
    """Return protein sequence for an Ensembl transcript."""
    url = f"{ENSEMBL_REST}/sequence/id/{transcript_id}"
    params = {'type': 'protein', 'content-type': 'text/plain'}
    try:
        r = requests.get(url, params=params, timeout=timeout)
        r.raise_for_status()
        return r.text.strip()
    except Exception:
        return ''

def extract_genomic_context(seq: str, pos_1based: int, k: int = 5) -> dict:
    """
    Extract ±k nucleotide neighbourhood around a 1-based CDS position.
    Returns individual upstream/downstream columns.
    """
    idx = pos_1based - 1  # 0-based
    upstream   = seq[max(0, idx - k): idx].rjust(k, 'N')[:k]
    downstream = seq[idx + 1: idx + 1 + k].ljust(k, 'N')[:k]
    result = {}
    for i, nuc in enumerate(upstream, 1):
        result[f'nuc_up_{i}'] = nuc
    for i, nuc in enumerate(downstream, 1):
        result[f'nuc_dn_{i}'] = nuc
    return result

def extract_protein_context(prot_seq: str, pos_1based: int, k: int = 5) -> dict:
    """
    Extract ±k amino-acid neighbourhood around a 1-based protein position.
    """
    idx = pos_1based - 1
    upstream   = prot_seq[max(0, idx - k): idx].rjust(k, 'X')[:k]
    downstream = prot_seq[idx + 1: idx + 1 + k].ljust(k, 'X')[:k]
    result = {}
    for i, aa in enumerate(upstream, 1):
        result[f'aa_up_{i}'] = aa
    for i, aa in enumerate(downstream, 1):
        result[f'aa_dn_{i}'] = aa
    return result

print("Context extraction helpers defined.")
```

```python
# ── Gene → canonical transcript mapping via Ensembl ─────────────────────────
def get_canonical_transcript(gene_symbol: str) -> dict:
    """Return canonical transcript ID and protein ID for a HGNC gene symbol."""
    url = f"{ENSEMBL_REST}/lookup/symbol/homo_sapiens/{gene_symbol}"
    headers = {'Content-Type': 'application/json'}
    try:
        r = requests.get(url, headers=headers,
                         params={'expand': 1, 'content-type': 'application/json'},
                         timeout=20)
        r.raise_for_status()
        data = r.json()
        for t in data.get('Transcript', []):
            if t.get('is_canonical') == 1:
                return {'transcript': t['id'], 'protein': t.get('Translation', {}).get('id')}
    except Exception:
        pass
    return {}

# Cache transcripts for all genes to avoid repeated API calls
TRANSCRIPT_CACHE_FILE = RAW_DIR / "transcript_cache.json"
if TRANSCRIPT_CACHE_FILE.exists():
    transcript_cache = json.loads(TRANSCRIPT_CACHE_FILE.read_text())
else:
    transcript_cache = {}

unique_genes = combined['gene'].dropna().unique()
for gene in tqdm(unique_genes, desc="Fetching canonical transcripts"):
    if gene not in transcript_cache:
        transcript_cache[gene] = get_canonical_transcript(gene)
        time.sleep(0.15)

TRANSCRIPT_CACHE_FILE.write_text(json.dumps(transcript_cache, indent=2))
print(f"Transcript cache: {len(transcript_cache)} genes.")
```

```python
# ── Sequence cache: store CDS + protein sequences per transcript ─────────────
SEQ_CACHE_FILE = RAW_DIR / "sequence_cache.json"
if SEQ_CACHE_FILE.exists():
    seq_cache = json.loads(SEQ_CACHE_FILE.read_text())
else:
    seq_cache = {}

for gene, tinfo in tqdm(transcript_cache.items(), desc="Fetching sequences"):
    tid = tinfo.get('transcript')
    if tid and tid not in seq_cache:
        seq_cache[tid] = {
            'cds':     get_transcript_cds(tid),
            'protein': get_transcript_protein(tid)
        }
        time.sleep(0.2)

SEQ_CACHE_FILE.write_text(json.dumps(seq_cache, indent=2))
print(f"Sequence cache: {len(seq_cache)} transcripts.")
```

```python
# ── Apply context extraction to each variant ─────────────────────────────────
def get_context_features(row) -> dict:
    gene     = row.get('gene')
    cds_pos  = row.get('cds_pos')
    prot_pos = row.get('protein_pos')
    tid      = transcript_cache.get(gene, {}).get('transcript')
    seqs     = seq_cache.get(tid, {})

    result = {}
    if seqs.get('cds') and cds_pos:
        result.update(extract_genomic_context(seqs['cds'], int(cds_pos)))
    if seqs.get('protein') and prot_pos:
        result.update(extract_protein_context(seqs['protein'], int(prot_pos)))
    return result

tqdm.pandas(desc="Extracting sequence context")
ctx_feats = combined.progress_apply(get_context_features, axis=1)
ctx_df    = pd.DataFrame(ctx_feats.tolist())
combined  = pd.concat([combined.reset_index(drop=True),
                        ctx_df.reset_index(drop=True)], axis=1)

nuc_ctx_cols = [f'nuc_up_{i}' for i in range(1,6)] + [f'nuc_dn_{i}' for i in range(1,6)]
aa_ctx_cols  = [f'aa_up_{i}'  for i in range(1,6)] + [f'aa_dn_{i}'  for i in range(1,6)]
print("Context features added:", nuc_ctx_cols + aa_ctx_cols)
```

### 5.3 Biochemical & Structural Effects

```python
def biochemical_features(row) -> dict:
    r1 = row.get('ref_aa1') or 'X'
    a1 = row.get('alt_aa1') or 'X'
    return {
        # Individual physicochemical values
        'ref_hydrophobicity': AA_HYDROPHOBICITY.get(r1, 0),
        'alt_hydrophobicity': AA_HYDROPHOBICITY.get(a1, 0),
        'delta_hydrophobicity': AA_HYDROPHOBICITY.get(a1, 0) - AA_HYDROPHOBICITY.get(r1, 0),

        'ref_polarity': AA_POLARITY.get(r1, 0),
        'alt_polarity': AA_POLARITY.get(a1, 0),
        'polarity_change': int(AA_POLARITY.get(r1, 0) != AA_POLARITY.get(a1, 0)),

        'ref_mw': AA_MW.get(r1, 110),
        'alt_mw': AA_MW.get(a1, 110),
        'delta_mw': AA_MW.get(a1, 110) - AA_MW.get(r1, 110),

        'ref_charge': AA_CHARGE.get(r1, 0),
        'alt_charge': AA_CHARGE.get(a1, 0),
        'charge_change': AA_CHARGE.get(a1, 0) - AA_CHARGE.get(r1, 0),

        # Proline / Cysteine flags (known structural disruptors)
        'alt_is_pro': int(a1 == 'P'),
        'ref_is_cys': int(r1 == 'C'),
        'alt_is_cys': int(a1 == 'C'),

        # Grantham distance (biochemical distance between AAs)
        # Simplified Grantham proxy = |Δhydrophobicity| + |Δmw|/10 + |Δcharge|*10
        'grantham_proxy': (
            abs(AA_HYDROPHOBICITY.get(a1,0) - AA_HYDROPHOBICITY.get(r1,0)) +
            abs(AA_MW.get(a1,110) - AA_MW.get(r1,110)) / 10 +
            abs(AA_CHARGE.get(a1,0) - AA_CHARGE.get(r1,0)) * 10
        )
    }

tqdm.pandas(desc="Biochemical features")
bio_feats = combined.progress_apply(biochemical_features, axis=1)
bio_df    = pd.DataFrame(bio_feats.tolist())
combined  = pd.concat([combined.reset_index(drop=True),
                        bio_df.reset_index(drop=True)], axis=1)
print("Biochemical features added:", bio_df.columns.tolist())
```

### 5.4 Evolutionary Conservation Scores (GERP, PhyloP, PhastCons)

```python
# ── Fetch conservation scores via Ensembl VEP REST ──────────────────────────
# VEP returns phyloP, phastCons, GERP++ for each variant.
# We query in batches of 200 (API limit).

VEP_REST = "https://rest.ensembl.org/vep/human/hgvs"

def vep_batch(hgvs_list: list, batch_size: int = 200) -> dict:
    """
    Query Ensembl VEP REST for a list of HGVS notations.
    Returns {hgvs: {feature: value}} dict.
    """
    results = {}
    headers = {'Content-Type': 'application/json',
               'Accept':       'application/json'}
    for i in range(0, len(hgvs_list), batch_size):
        batch = [h for h in hgvs_list[i:i+batch_size] if isinstance(h, str)]
        if not batch:
            continue
        payload = json.dumps({'hgvs_notations': batch})
        for attempt in range(3):
            try:
                r = requests.post(VEP_REST, headers=headers,
                                  data=payload, timeout=60)
                r.raise_for_status()
                for entry in r.json():
                    hgvs = entry.get('input')
                    tc   = (entry.get('transcript_consequences') or [{}])[0]
                    co   = entry.get('colocated_variants', [{}])[0]
                    results[hgvs] = {
                        # Conservation
                        'phylop_score':    entry.get('phylop'),
                        'phastcons_score': entry.get('phastcons'),
                        'gerp_score':      entry.get('gerp'),
                        # From transcript consequence
                        'sift_score':    tc.get('sift_score'),
                        'sift_pred':     tc.get('sift_prediction'),
                        'polyphen_score': tc.get('polyphen_score'),
                        'polyphen_pred':  tc.get('polyphen_prediction'),
                        'cadd_phred':    tc.get('cadd_phred'),
                        'cadd_raw':      tc.get('cadd_raw'),
                        # Population (MAF)
                        'maf_global':    co.get('minor_allele_freq'),
                    }
                break
            except Exception as e:
                print(f"  VEP retry {attempt+1}: {e}")
                time.sleep(2 ** attempt)
        time.sleep(0.5)
    return results

print("VEP batch helper defined.")
```

```python
# Build HGVS list for VEP: prefer genomic HGVS or use transcript HGVS
# For competition data, use transcript-based HGVS (hgvs_c)
VEP_CACHE_FILE = RAW_DIR / "vep_cache.json"
if VEP_CACHE_FILE.exists():
    vep_cache = json.loads(VEP_CACHE_FILE.read_text())
else:
    vep_cache = {}

hgvs_to_query = [
    h for h in combined['hgvs_c'].dropna().unique()
    if h not in vep_cache
]
print(f"Querying VEP for {len(hgvs_to_query):,} variants …")

new_vep = vep_batch(hgvs_to_query)
vep_cache.update(new_vep)
VEP_CACHE_FILE.write_text(json.dumps(vep_cache, indent=2))
print(f"VEP cache size: {len(vep_cache):,}")

# Map VEP scores back to combined dataframe
vep_score_df = combined['hgvs_c'].map(vep_cache).apply(
    lambda x: x if isinstance(x, dict) else {}
).apply(pd.Series)

# Rename to avoid collision with gnomAD scores already present
for col in ['sift_score','sift_pred','polyphen_score','polyphen_pred']:
    if col in combined.columns and col in vep_score_df.columns:
        vep_score_df.rename(columns={col: f'{col}_vep'}, inplace=True)

combined = pd.concat([combined.reset_index(drop=True),
                       vep_score_df.reset_index(drop=True)], axis=1)
print("Conservation/VEP features added.")
```

### 5.5 Population Frequencies (gnomAD per-population)

```python
# ── Fetch per-population AFs from gnomAD v4 ──────────────────────────────────
GNOMAD_VARIANT_QUERY = """
query VariantAF($variantId: String!, $dataset: DatasetId!) {
  variant(variantId: $variantId, dataset: $dataset) {
    genome {
      ac af an
      populations { id ac af an }
    }
    exome {
      ac af an
      populations { id ac af an }
    }
  }
}
"""

POPULATION_IDS = ['afr','amr','asj','eas','fin','mid','nfe','sas','remaining']

POP_CACHE_FILE = RAW_DIR / "pop_af_cache.json"
if POP_CACHE_FILE.exists():
    pop_cache = json.loads(POP_CACHE_FILE.read_text())
else:
    pop_cache = {}

def fetch_pop_af(variant_id: str) -> dict:
    """variant_id format: chrom-pos-ref-alt (e.g. 17-43057051-G-A)"""
    try:
        r = requests.post(
            GNOMAD_API,
            json={'query': GNOMAD_VARIANT_QUERY,
                  'variables': {'variantId': variant_id, 'dataset': 'gnomad_r4'}},
            timeout=30
        )
        r.raise_for_status()
        data = r.json().get('data', {}).get('variant') or {}
        result = {}
        for src in ['genome', 'exome']:
            src_data = data.get(src) or {}
            result[f'af_{src}'] = src_data.get('af')
            result[f'ac_{src}'] = src_data.get('ac')
            result[f'an_{src}'] = src_data.get('an')
            for pop in src_data.get('populations', []):
                pid = pop['id'].lower()
                result[f'af_{src}_{pid}'] = pop.get('af')
        return result
    except Exception:
        return {}

# Only fetch for variants that have a gnomAD variant_id
ids_to_fetch = [
    vid for vid in combined['variant_id'].dropna().unique()
    if vid not in pop_cache
]
print(f"Fetching population AFs for {len(ids_to_fetch):,} variants …")
for vid in tqdm(ids_to_fetch):
    pop_cache[vid] = fetch_pop_af(vid)
    time.sleep(0.3)

POP_CACHE_FILE.write_text(json.dumps(pop_cache, indent=2))

pop_df = combined['variant_id'].map(pop_cache).apply(
    lambda x: x if isinstance(x, dict) else {}
).apply(pd.Series)

combined = pd.concat([combined.reset_index(drop=True),
                       pop_df.reset_index(drop=True)], axis=1)
print("Population AF features added.")
```

### 5.6 In Silico Risk Scores (CADD, REVEL, AlphaMissense, VEST4, MutPred2)

```python
# ── CADD scores via API (cadd.gs.washington.edu) ─────────────────────────────
CADD_API = "https://cadd.gs.washington.edu/api/v1.0/"

def fetch_cadd_scores(variant_ids: list,
                      genome: str = 'GRCh38-v1.7') -> dict:
    """
    Fetch CADD PHRED and raw scores.
    variant_ids: list of 'chrom-pos-ref-alt' strings
    """
    results = {}
    for vid in tqdm(variant_ids, desc="CADD"):
        try:
            chrom, pos, ref, alt = vid.split('-')
            url = f"{CADD_API}{genome}/{chrom}:{pos}_{ref}_{alt}"
            r = requests.get(url, timeout=20)
            if r.status_code == 200:
                rows = r.json()
                if rows:
                    results[vid] = {
                        'cadd_raw':   rows[0].get('RawScore'),
                        'cadd_phred': rows[0].get('PHRED')
                    }
        except Exception:
            pass
        time.sleep(0.2)
    return results

# ── REVEL scores (static file lookup) ───────────────────────────────────────
# REVEL provides a pre-computed score file; load it if available.
REVEL_FILE = RAW_DIR / "revel_all_chromosomes.csv.gz"

def load_revel(path: Path) -> pd.DataFrame:
    if not path.exists():
        print(f"[WARN] REVEL file not found at {path}. Download from:"
              " https://sites.google.com/site/revelgenomics/downloads")
        return pd.DataFrame()
    print("Loading REVEL scores …")
    revel = pd.read_csv(path, compression='gzip', low_memory=False)
    revel.columns = revel.columns.str.strip().str.lower()
    return revel

revel_df = load_revel(REVEL_FILE)

# ── AlphaMissense scores (static lookup, downloaded from Zenodo) ─────────────
ALPHAMISSENSE_FILE = RAW_DIR / "AlphaMissense_hg38.tsv.gz"

def load_alphamissense(path: Path) -> pd.DataFrame:
    if not path.exists():
        print(f"[WARN] AlphaMissense file not found at {path}. Download from:"
              " https://zenodo.org/record/8208688")
        return pd.DataFrame()
    print("Loading AlphaMissense scores …")
    am = pd.read_csv(path, sep='\t', compression='gzip',
                      comment='#', low_memory=False)
    am.columns = am.columns.str.strip().str.lower()
    return am

am_df = load_alphamissense(ALPHAMISSENSE_FILE)
print("In silico score helpers ready.")
```

```python
# ── Merge REVEL scores ───────────────────────────────────────────────────────
if not revel_df.empty:
    # REVEL uses chr, hg19_pos, ref, alt, aaref, aaalt, REVEL
    # Map on protein change if genomic coords are hidden
    revel_merge = revel_df[['aaref','aaalt','revel']].drop_duplicates()
    revel_merge.rename(columns={'aaref':'ref_aa1','aaalt':'alt_aa1',
                                 'revel':'revel_score'}, inplace=True)
    combined = combined.merge(revel_merge, on=['ref_aa1','alt_aa1'],
                               how='left')
    print("REVEL scores merged.")

# ── Merge AlphaMissense scores ────────────────────────────────────────────────
if not am_df.empty:
    # AlphaMissense cols: uniprot_id, transcript_id, protein_variant, am_pathogenicity, am_class
    am_merge = am_df[['transcript_id','protein_variant',
                       'am_pathogenicity','am_class']].copy()
    # protein_variant format: A123V (1-letter)
    am_merge['ref_aa1']    = am_merge['protein_variant'].str[0]
    am_merge['alt_aa1']    = am_merge['protein_variant'].str[-1]
    am_merge['protein_pos'] = am_merge['protein_variant'].str[1:-1].astype(float)
    am_merge.drop(columns='protein_variant', inplace=True)

    combined = combined.merge(
        am_merge[['ref_aa1','alt_aa1','protein_pos','am_pathogenicity','am_class']],
        on=['ref_aa1','alt_aa1','protein_pos'],
        how='left'
    )
    print("AlphaMissense scores merged.")
```

## 6. Data Cleaning

```python
# ── Define the final feature columns ────────────────────────────────────────

# Sequence / change
SEQ_COLS = ['ref_nuc', 'alt_nuc', 'cds_pos',
            'ref_aa1', 'alt_aa1', 'ref_aa3', 'alt_aa3', 'protein_pos']

# Local context
NUC_CTX_COLS = [f'nuc_up_{i}' for i in range(1,6)] + [f'nuc_dn_{i}' for i in range(1,6)]
AA_CTX_COLS  = [f'aa_up_{i}'  for i in range(1,6)] + [f'aa_dn_{i}'  for i in range(1,6)]

# Biochemical
BIO_COLS = [
    'ref_hydrophobicity','alt_hydrophobicity','delta_hydrophobicity',
    'ref_polarity','alt_polarity','polarity_change',
    'ref_mw','alt_mw','delta_mw',
    'ref_charge','alt_charge','charge_change',
    'alt_is_pro','ref_is_cys','alt_is_cys','grantham_proxy'
]

# Evolutionary conservation
CONS_COLS = ['phylop_score','phastcons_score','gerp_score']

# Population
POP_COLS = [
    'af_gnomad','ac_gnomad','an_gnomad',
    'af_genome','af_exome',
    'maf_global'
] + [f'af_genome_{p}' for p in POPULATION_IDS] \
  + [f'af_exome_{p}'  for p in POPULATION_IDS]

# In silico scores
INSILICO_COLS = [
    'sift_score','polyphen_score',
    'cadd_raw','cadd_phred',
    'revel_score',
    'am_pathogenicity'
]

# Categorical in-silico predictions (for encoding)
CAT_PRED_COLS = [
    'sift_pred','polyphen_pred','am_class',
    'ref_nuc','alt_nuc','ref_aa1','alt_aa1',
    'ref_aa3','alt_aa3'
] + NUC_CTX_COLS + AA_CTX_COLS

ALL_FEATURE_COLS = (SEQ_COLS + NUC_CTX_COLS + AA_CTX_COLS +
                    BIO_COLS + CONS_COLS + POP_COLS + INSILICO_COLS)

META_COLS = ['gene', 'hgvs_c', 'hgvs_p', 'label']

# Keep only columns that exist
available_feat = [c for c in ALL_FEATURE_COLS if c in combined.columns]
available_cat  = [c for c in CAT_PRED_COLS    if c in combined.columns]

print(f"Feature columns available: {len(available_feat)} / {len(ALL_FEATURE_COLS)}")
```

```python
# ── Build cleaned feature matrix ─────────────────────────────────────────────
keep_cols = META_COLS + available_feat + available_cat
keep_cols = [c for c in keep_cols if c in combined.columns]

df_clean = combined[keep_cols].copy()

# ── 1. Drop rows with no amino-acid change identified ────────────────────────
df_clean.dropna(subset=['ref_aa1','alt_aa1'], inplace=True)
print(f"After AA filter: {len(df_clean):,}")

# ── 2. Remove synonymous / stop-gained variants that slipped through ─────────
df_clean = df_clean[df_clean['ref_aa1'] != df_clean['alt_aa1']]
df_clean = df_clean[~df_clean['alt_aa1'].isin(['*', 'Ter'])]
print(f"After synonymous/stop filter: {len(df_clean):,}")

# ── 3. Impute missing numeric scores with column median ───────────────────────
numeric_cols = [c for c in available_feat if c not in available_cat]
numeric_cols = [c for c in numeric_cols if df_clean[c].dtype in
                ['float64','float32','int64','int32']]

medians = df_clean[numeric_cols].median()
df_clean[numeric_cols] = df_clean[numeric_cols].fillna(medians)

# ── 4. Fill missing categoricals with 'unknown' ───────────────────────────────
for col in available_cat:
    df_clean[col] = df_clean[col].fillna('unknown').astype(str).str.lower()

# ── 5. Clamp extreme allele frequencies to [0, 1] ────────────────────────────
af_cols = [c for c in df_clean.columns if c.startswith('af_')]
df_clean[af_cols] = df_clean[af_cols].clip(0, 1)

# ── 6. Remove duplicates on (gene, hgvs_p) ───────────────────────────────────
df_clean.drop_duplicates(subset=['gene','hgvs_p'], keep='first', inplace=True)
df_clean.reset_index(drop=True, inplace=True)
print(f"After dedup: {len(df_clean):,}")
print(df_clean['label'].value_counts().rename({0:'Benign', 1:'Pathogenic'}))
```

```python
# ── 7. Encode categorical columns ────────────────────────────────────────────
# One-hot encode prediction labels; label-encode single-character AAs/nucs
from sklearn.preprocessing import LabelEncoder

# Binary predictions → numeric
pred_map = {
    'sift_pred':     {'tolerated': 0, 'deleterious': 1, 'unknown': np.nan},
    'polyphen_pred': {'benign': 0, 'possibly_damaging': 0.5, 'probably_damaging': 1,
                      'unknown': np.nan},
    'am_class':      {'likely_benign': 0, 'ambiguous': 0.5,
                      'likely_pathogenic': 1, 'unknown': np.nan}
}
for col, mapping in pred_map.items():
    if col in df_clean.columns:
        df_clean[f'{col}_num'] = df_clean[col].map(mapping)

# Ordinal-encode single-letter amino-acids and nucleotides
single_char_cats = [c for c in available_cat if c in
                    ['ref_aa1','alt_aa1','ref_nuc','alt_nuc'] +
                    NUC_CTX_COLS + AA_CTX_COLS]
le = LabelEncoder()
for col in single_char_cats:
    if col in df_clean.columns:
        df_clean[f'{col}_enc'] = le.fit_transform(df_clean[col].astype(str))

print("Encoding complete.")
```

```python
# ── 8. Quality summary ────────────────────────────────────────────────────────
print("=" * 60)
print("DATASET QUALITY REPORT")
print("=" * 60)
print(f"Total variants   : {len(df_clean):,}")
print(f"Pathogenic (1)   : {(df_clean.label==1).sum():,}")
print(f"Benign     (0)   : {(df_clean.label==0).sum():,}")
print(f"Features         : {df_clean.shape[1]}")
print("\nMissing values per feature (top 20):")
mv = df_clean.isna().sum().sort_values(ascending=False)
print(mv[mv > 0].head(20).to_string())
```

## 7. Panel-Specific Subsetting & Train/Test Split

```python
def balanced_sample(df: pd.DataFrame,
                    n_path: int,
                    n_benign: int,
                    seed: int = SEED) -> pd.DataFrame:
    """Return a balanced subset with exactly n_path Pathogenic & n_benign Benign."""
    path_df  = df[df.label == 1].sample(min(n_path,  (df.label==1).sum()),
                                         random_state=seed)
    benign_df = df[df.label == 0].sample(min(n_benign, (df.label==0).sum()),
                                          random_state=seed)
    return pd.concat([path_df, benign_df]).sample(frac=1, random_state=seed)

def make_panel(df: pd.DataFrame, genes: list,
               train_n: tuple, test_n: tuple,
               name: str, seed: int = SEED):
    """Subset to gene panel, then produce balanced train/test."""
    panel = df[df['gene'].isin(genes)].copy()
    print(f"\n{name}: {len(panel):,} total variants in panel genes")
    train = balanced_sample(panel, train_n[0], train_n[1], seed)
    # Remove train rows before sampling test to prevent leakage
    remainder = panel.drop(train.index)
    test  = balanced_sample(remainder, test_n[0], test_n[1], seed)
    print(f"  Train: {len(train):,} | Test: {len(test):,}")
    return train, test


# 1. General Dataset
train_general, test_general = make_panel(
    df_clean, df_clean['gene'].unique().tolist(),
    train_n=(1500, 1500), test_n=(1000, 1000),
    name='General'
)

# 2. Hereditary Cancer Panel
train_cancer, test_cancer = make_panel(
    df_clean, HEREDITARY_CANCER_GENES,
    train_n=(200, 200), test_n=(100, 100),
    name='Hereditary Cancer'
)

# 3. PAH Gene Panel
train_pah, test_pah = make_panel(
    df_clean, PAH_GENES,
    train_n=(200, 200), test_n=(100, 100),
    name='PAH (Phenylketonuria)'
)

# 4. CFTR Gene Panel
train_cftr, test_cftr = make_panel(
    df_clean, CFTR_GENES,
    train_n=(70, 70), test_n=(30, 30),
    name='CFTR (Cystic Fibrosis)'
)
```

## 8. Hide Genomic Coordinates for Competition Delivery

```python
# Columns to STRIP before delivering to competitors
# (chromosome, position, variant IDs that link to external DBs)
HIDDEN_COLS = [
    'Chromosome','Start','Stop','PositionVCF',
    'ChromosomeAccession','variant_id',
    'allele_id','RCVaccession',
    'dedup_key','hgvs_c','hgvs_p',    # HGVS also reveals genomic location
    'Name',                            # raw ClinVar name has chr/pos
    'RS# (dbSNP)'
]

def strip_hidden(df: pd.DataFrame) -> pd.DataFrame:
    drop = [c for c in HIDDEN_COLS if c in df.columns]
    return df.drop(columns=drop)

# Also hide labels from the test set
def make_unlabelled(df: pd.DataFrame) -> pd.DataFrame:
    return strip_hidden(df).drop(columns=['label'], errors='ignore')

# Labelled train sets (for competitors after passing report stage)
datasets = {
    'general':  (train_general, test_general),
    'cancer':   (train_cancer,  test_cancer),
    'pah':      (train_pah,     test_pah),
    'cftr':     (train_cftr,    test_cftr),
}

for panel_name, (train, test) in datasets.items():
    out_dir = PROC_DIR / panel_name
    out_dir.mkdir(parents=True, exist_ok=True)

    # Labelled training set
    strip_hidden(train).to_csv(out_dir / 'train_labelled.csv', index=False)

    # Unlabelled test set (for release during competition)
    make_unlabelled(test).to_csv(out_dir / 'test_unlabelled.csv', index=False)

    # Labelled test set (ground truth, kept secret by organiser)
    strip_hidden(test).to_csv(out_dir / 'test_groundtruth.csv', index=False)

    print(f"{panel_name:12s} → {out_dir}  "
          f"(train={len(train)}, test={len(test)})")

print("\nAll competition datasets saved.")
```

---
## 9. Final Column Anonymisation (Remove column names)

```python
# Per the competition spec, column names are NOT disclosed.
# Deliver CSV files with numeric column headers.

# Keep a private mapping file for the organiser.
for panel_name in datasets:
    out_dir = PROC_DIR / panel_name
    for fname in ['train_labelled.csv', 'test_unlabelled.csv']:
        fpath = out_dir / fname
        if not fpath.exists():
            continue
        df_tmp = pd.read_csv(fpath)
        col_mapping = {c: f'F{i:03d}' for i, c in enumerate(df_tmp.columns)}
        # Save mapping (secret)
        pd.DataFrame.from_dict(col_mapping, orient='index',
                                columns=['anon_name']).to_csv(
            out_dir / f'{fname.replace(".csv", "_column_map.csv")}')
        # Rename and save anonymous version
        df_tmp.rename(columns=col_mapping).to_csv(
            out_dir / fname.replace('.csv', '_anonymous.csv'), index=False)

print("Anonymous datasets (no column names) saved.")
```
