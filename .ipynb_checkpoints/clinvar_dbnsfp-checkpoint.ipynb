```python
from pathlib import Path

BASE = Path.cwd()
DATA = BASE / "data"
RAW = DATA / "raw"
INTERIM = DATA / "interim"

for p in [DATA, RAW, INTERIM]:
    p.mkdir(parents=True, exist_ok=True)

print("Folders created.")
```

```python
import pandas as pd
import numpy as np
import re
from pathlib import Path

pd.set_option("display.max_columns", 200)
pd.set_option("display.max_colwidth", 200)

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
```

```python
needed_clinvar_columns = [
    "VariationID", "GeneSymbol", "Type", "Assembly", "OriginSimple",
    "ClinicalSignificance", "ReviewStatus", "Chromosome", "Start",
    "ReferenceAllele", "AlternateAllele", "ReferenceAlleleVCF",
    "AlternateAlleleVCF", "PositionVCF", "Name", "label"  # Added 'label' here
]
```

```python
clinvar_path = INTERIM / "clinvar_1.csv"

clinvar = pd.read_csv(clinvar_path, usecols=needed_clinvar_columns, low_memory=False)

print(clinvar.shape)
clinvar.head(3)
```

```python
chromosomes = clinvar['Chromosome'].unique()
print(chromosomes)
```

```python
dbnsfp_cols = [
    "#chr", "pos(1-based)", "ref", "alt",
    "aaref", "aaalt", "aapos", "genename",
    "Ensembl_transcriptid", "HGVSc_snpEff", "HGVSp_snpEff",
    "SIFT_score", "SIFT_converted_rankscore",
    "Polyphen2_HDIV_score", "Polyphen2_HDIV_rankscore",
    "REVEL_score", "REVEL_rankscore",
    "CADD_phred", "CADD_raw_rankscore",
    "AlphaMissense_score", "AlphaMissense_rankscore",
    "MetaRNN_score", "MetaRNN_rankscore",
    "ClinPred_score", "ClinPred_rankscore",
    "BayesDel_addAF_score", "BayesDel_addAF_rankscore",
    "phyloP100way_vertebrate", "phastCons100way_vertebrate",
    "GERP++_RS", "GERP++_RS_rankscore",
    "gnomAD4.1_joint_AF", "gnomAD4.1_joint_AC", "gnomAD4.1_joint_AN",
    "dbNSFP_POPMAX_AF"
]
```

```python
# Standardize ClinVar for merging
print("Preparing ClinVar data...")

# Ensure Chromosome is string and without 'chr' prefix
clinvar['Chromosome'] = clinvar['Chromosome'].astype(str).str.replace('chr', '', case=False)

# Ensure Position is integer
clinvar['PositionVCF'] = pd.to_numeric(clinvar['PositionVCF'], errors='coerce')
clinvar = clinvar.dropna(subset=['PositionVCF'])
clinvar['PositionVCF'] = clinvar['PositionVCF'].astype(int)

print(f"ClinVar ready. Total variants to process: {len(clinvar)}")
```

```python
from tqdm import tqdm
import os

merged_chunks = []

base_path = "D:/datasets/dbNSFP5.3.1a/dbNSFP5.3.1a/"

for chr_id in chromosomes:
    # Filter ClinVar subset for current chromosome
    cv_sub = clinvar[clinvar['Chromosome'] == chr_id].copy()
    if cv_sub.empty:
        continue

    file_path = os.path.join(base_path, f"dbNSFP5.3.1a_variant.chr{chr_id}.gz")
    
    if not os.path.exists(file_path):
        print(f"Warning: File for Chr {chr_id} not found.")
        continue

    # Use chunksize to read small portions of the massive file at a time
    reader = pd.read_csv(
        file_path, 
        sep="\t", 
        compression="gzip", 
        usecols=dbnsfp_cols,
        chunksize=200000, 
        low_memory=True
    )

    # Process each chunk
    for chunk in tqdm(reader, desc=f"Processing Chr {chr_id}", unit="chunk"):
        
        # 1. Force dbNSFP types to match ClinVar exactly
        # Use 'str' for Chromosome and 'int' for Position
        chunk['#chr'] = chunk['#chr'].astype(str).str.replace('chr', '', case=False)
        chunk['pos(1-based)'] = pd.to_numeric(chunk['pos(1-based)'], errors='coerce').fillna(0).astype(int)
        
        # 2. Ensure ClinVar subset is also explicitly typed (redundant but safe)
        cv_sub['Chromosome'] = cv_sub['Chromosome'].astype(str)
        cv_sub['PositionVCF'] = cv_sub['PositionVCF'].astype(int)
        
        # 3. Perform the Merge
        matched = pd.merge(
            cv_sub, 
            chunk, 
            left_on=["Chromosome", "PositionVCF", "ReferenceAlleleVCF", "AlternateAlleleVCF"], 
            right_on=["#chr", "pos(1-based)", "ref", "alt"], 
            how="inner" 
        )
        
        if not matched.empty:
            matched = matched.drop(columns=["#chr", "pos(1-based)", "ref", "alt"])
            merged_chunks.append(matched)

print("\n--- All Chromosomes Processed ---")
```

```python
if merged_chunks:
    print("Finalizing dataframe assembly...")
    # Combine all found matches
    dbnsfp_matches = pd.concat(merged_chunks, ignore_index=True)
    
    # Left join back to original ClinVar to keep variants without any dbNSFP scores
    final_df = pd.merge(clinvar, dbnsfp_matches, on=needed_clinvar_columns, how="left")
    
    # Save the result
    output_name = "clinvar_dbnsfp_merged.parquet"
    final_df.to_parquet(output_name, index=False)
    
    print(f"Success! Final file saved as {output_name}")
    print(f"Final Row Count: {len(final_df)}")
else:
    print("No matches were found. Please check your file paths and join keys.")
```
