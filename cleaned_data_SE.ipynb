```python
mw = {
"A": 89.1, "R": 174.2, "N": 132.1, "D": 133.1,
"C": 121.2, "E": 147.1, "Q": 146.2, "G": 75.1,
"H": 155.2, "I": 131.2, "L": 131.2, "K": 146.2,
"M": 149.2, "F": 165.2, "P": 115.1, "S": 105.1,
"T": 119.1, "W": 204.2, "Y": 181.2, "V": 117.1
}
```

```python
hydro = {
"A": 1.8, "R": -4.5, "N": -3.5, "D": -3.5,
"C": 2.5, "E": -3.5, "Q": -3.5, "G": -0.4,
"H": -3.2, "I": 4.5, "L": 3.8, "K": -3.9,
"M": 1.9, "F": 2.8, "P": -1.6, "S": -0.8,
"T": -0.7, "W": -0.9, "Y": -1.3, "V": 4.2
}
```

```python
polarity = {
"A": 0, "V": 0, "L": 0, "I": 0, "M": 0, "F": 0, "W": 0, "P": 0,   # nonpolar
"G": 1, "S": 1, "T": 1, "Y": 1, "C": 1,                           # polar
"D": 2, "E": 2, "K": 2, "R": 2, "H": 2, "N": 2, "Q": 2            # charged
}
```

```python
df["mw_change"] = df["aaalt"].map(mw) - df["aaref"].map(mw)
df["hydro_change"] = df["aaalt"].map(hydro) - df["aaref"].map(hydro)
df["polarity_change"] = df["aaalt"].map(polarity) - df["aaref"].map(polarity)
df.head()
```

```python
cols_split = [
"aapos", "genename"
    
]

for col in cols_split:
    df[col] = df[col].astype(str).str.split(";").str[0]
    
df.head()
```

```python
import re

def extract_first_float(x):
    if pd.isna(x):
        return None

    # Find all numbers like 0.123 inside the string
    matches = re.findall(r'\d*\.\d+', str(x))

    if len(matches) > 0:
        return float(matches[0])  # take first valid number

    return None

cols = ['REVEL_score', 'AlphaMissense_score', 'SIFT_score', 'MetaRNN_score']

for col in cols:
    df[col] = df[col].apply(extract_first_float)
    
    
df.head()
```

```python
import re

def extract_max_float(x):
    if pd.isna(x):
        return None

    matches = re.findall(r'\d*\.\d+|\d+', str(x))  # handles 1.0 and 1

    if matches:
        return max([float(m) for m in matches])

    return None

df['Polyphen2_HDIV_score'] = df['Polyphen2_HDIV_score'].apply(extract_max_float)
```

```python
numeric_cols = [
    'REVEL_score',
    'AlphaMissense_score',
    'SIFT_score',
    'gnomAD4.1_joint_POPMAX_AF',
    'dbNSFP_POPMAX_AF',
    'MetaRNN_score',
    'CADD_phred',
    'ClinPred_score',
    'BayesDel_addAF_score',
    'M-CAP_score',
    'PrimateAI_score',
    'GERP++_RS',
    'phyloP100way_vertebrate',
    'phastCons100way_vertebrate',
    'gnomAD4.1_joint_AF',
    'Polyphen2_HDIV_score'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```
