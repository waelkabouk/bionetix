```python
import pandas as pd
from pyfaidx import Fasta

clin_nsfp = pd.read_csv(r'C:\Users\المنارة البيضاء\bionetix\data\interim\nsfp_cleaned.csv')

genome = Fasta(r'C:\Users\المنارة البيضاء\bionetix\data\hg38\hg38.fa')

def get_context(row, window=5):
    chrom = f"chr{row['#chr']}"
    pos   = row['pos(1-based)']
    idx   = pos - 1
    
    seq = str(genome[chrom][idx - window : idx + window + 1])
    
    return pd.Series({
        'nuc_before' : seq[0:window],
        'nuc_ref'    : seq[window],
        'nuc_after'  : seq[window+1:]
    })

clin_nsfp[['nuc_before', 'nuc_ref', 'nuc_after']] = clin_nsfp.apply(
    get_context, axis=1
)

clin_nsfp.to_csv(
    r'C:\Users\المنارة البيضاء\bionetix\data\interim\nsfp_nuc_squences_added.csv',
    index=False
)
```
